```python
from google_play_scraper import reviews_all
import pandas as pd

# Mengambil review dari aplikasi Tokopedia
result = reviews_all(
    'com.tokopedia.tkpd',  # ID aplikasi, bisa ganti ke app lain
    lang='id',
    country='id'
)

# Simpan ke DataFrame
df = pd.DataFrame(result)[['content', 'score']]

# Labelkan sentimen berdasarkan skor bintang
def label_sentiment(score):
    if score >= 4:
        return 'positive'
    elif score <= 2:
        return 'negative'
    else:
        return 'neutral'

df['sentiment'] = df['score'].apply(label_sentiment)
df = df[['content', 'sentiment']]

# Ambil 3000 data acak
df = df.sample(n=3000, random_state=42).reset_index(drop=True)

# Simpan ke CSV
df.to_csv("google_play_tokopedia_3000.csv", index=False)
print("Selesai! File saved as google_play_tokopedia_3000.csv")
```
